```python
file_string = open('names.txt','r').read()
names = file_string.split('\n')
```

```python
characters_list = sorted(set(list(file_string)))
characters_list.remove('\n')
characters_list.append('.')
characters_list
```

```python
import torch
import torch.nn as nn

n = 4 #context size. n-1 is input context size and 1 is the output
input_context_size = n - 1
embed_dimensions = 50
no_of_characters = len(characters_list)
lookup_matrix = torch.randn((no_of_characters, embed_dimensions), requires_grad = True)
char_to_idx = {c: i for i, c in enumerate(characters_list)}
learning_rate = 0.01

lookup_matrix.shape
lookup_matrix[0]
```

```python
idx_to_char = {i: c for c, i in char_to_idx.items()}
```

```python
torch.cat((lookup_matrix[0], lookup_matrix[0], lookup_matrix[0]))
```

```python
# Neural network weights
hidden_layer_size = 200
fan_in = embed_dimensions * input_context_size

# You probably have questions about this magic 5/3 number and what the hell fan_in is.
# It is supposed to do with Kaiming shift during initialization so that the hidden layer
# output (i.e. output of tanh operation) is not saturated.
# With 150 inputs in W1 having a mean 0 and std 1, the output of the activation function (input @ W1)
# would have a std of 12.2 and tanh(12.2) is 1. So basically, you would get a unit row matrix and that screws up
# the entire thing. If you want to understand this better, replace the following 2 lines with
# W1 = torch.randn((fan_in, hidden_layer_size), requires_grad = True)
W1 = torch.randn((fan_in, hidden_layer_size)) * (5/3) / fan_in**0.5
W1.requires_grad = True
b1 = torch.zeros(hidden_layer_size, requires_grad=True)

# Since we are scaling W2 and b2 by 0.01, that operation turns them into non-leaf nodes.
# So, we first scale and then say `requires_grad = True`
W2 = torch.randn((hidden_layer_size, no_of_characters)) * 0.01
W2.requires_grad = True
b2 = torch.zeros(no_of_characters, requires_grad=True)

# Another way to do that is
# W2 = torch.randn((hidden_layer_size, no_of_characters), requires_grad=True)
# with torch.no_grad():
#    W2 *= 0.01

parameters = [lookup_matrix, W1, b1, W2, b2]

print(f"{W1.shape} {W2.shape}")
```

```python
W1_out_std = []
losses = []
```

```python
def forward_pass(input_context):
    input_context_embeddings = torch.cat(torch.unbind(lookup_matrix[[char_to_idx[c] for c in input_context]]))
    W1_out = torch.tanh(input_context_embeddings @ W1 + b1)
    W2_out = W1_out @ W2 + b2

    logits = W2_out - W2_out.max(dim=-1, keepdim=True).values
    counts = logits.exp()
    probs = counts / counts.sum(dim=-1, keepdim=True)

    ### DEBUG START
    # print(f"W1_out={W1_out}")
    #print(f"\n W1={W1}\n b1={b1}\n W2={W2}\n b2={b2}")
    #print(f"\n input_context_embeddings={input_context_embeddings}\n W1_out={W1_out}\n W2_out={W2_out}\n logits={logits}\n counts={counts}\n")

    W1_out_std.append((input_context_embeddings @ W1 + b1).std(dim=-1).item())

    ### DEBUG END

    return probs
```

```python
def calculate_loss (probs, correct_char):
    correct_char_prob = probs[char_to_idx[correct_char]]
    ll = torch.log(correct_char_prob)
    nll = -ll

    return nll
```

```python
def gradient_descent (loss):
    for p in parameters:
        p.grad = None

    loss.backward()

    ### DEBUG CODE START
    #for i, p in enumerate(parameters):
    #    print(f"param {i} grad norm: {p.grad.norm().item():.4f}")
    ### DEBUG CODE END

    for i, p in zip(range(len(parameters)), parameters):
        p.data += -(learning_rate * p.grad)
```

```python
import matplotlib.pyplot as plt

def visualize (values):
    plt.figure(figsize=(10, 5))
    plt.plot(values)
    plt.xlabel("Name (training step)")
    plt.ylabel("Mean loss")
    plt.title("Loss over training")
    plt.grid(True)
```

```python
def moving_average(losses, window=100):
    return [sum(losses[i:i+window]) / window 
            for i in range(len(losses) - window + 1)]
```

```python
import random
random.seed(42)

def train ():
    training_set = list(names)

    random.shuffle(training_set)

    ### DEBUG CODE START
    #print(f"char_to_idx{char_to_idx}")
    ### DEBUG CODE END
    
    for name in training_set:
        transformed_name = ("."*input_context_size) + name + "."
        idx = 0
        batched_loss = []
        while True:
            input_chars = list(transformed_name[idx: (idx + input_context_size)])
            correct_output_char = transformed_name[idx + input_context_size]

            probs = forward_pass(input_chars)

            ### DEBUG CODE START
            #print(f"probs={probs}")
            #print(f"correct_char={correct_output_char}")
            #if probs[char_to_idx[correct_output_char]] == 0:
            #   break
            ### DEBUG CODE END
            
            loss = calculate_loss(probs, correct_output_char)
            batched_loss.append(loss)
            #print(f"loss: {loss}")
            #gradient_descent(loss)
            
            if (correct_output_char == '.'):
                break
            idx+=1

        losses.append(torch.stack(batched_loss).mean().item())
        gradient_descent(torch.stack(batched_loss).mean())
        #print(f"{torch.stack(batched_loss).mean().item()}")
        first_loss = False

    visualize(moving_average(losses, window=100))
        
```

```python
train()
```

```python
print(f"First loss: {losses[0]}")
```

```python
def generate_name():
    context = ['.']*input_context_size
    gen_name = []
    with torch.no_grad():
        while True:
            gen_probs = forward_pass(context)
            gen_next_idx = torch.multinomial(gen_probs, num_samples=1).item()
            gen_next_char = idx_to_char[gen_next_idx]

            gen_name.append(gen_next_char)

            if gen_next_char == '.':
                break
            context = context[1:] + [gen_next_char]
    return ''.join(gen_name)
```

```python
for i in range(10):
    print(f"{generate_name()}")
```
